```julia
using Random, Statistics
Random.seed!(123)
n = 250
x = rand(0:999, n)
y = rand(0:999, n)

v_1 = y[2:end] .- x[1:end-1]
v_2 = x[1:end-2] .+ 2 .* x[2:end-1] .- x[3:end]
v_3 = sin.(y[1:end-1]) ./ cos.(x[2:end])
v_sum = sum(exp.(-x[2:end]) ./ (x[1:end-1] .+ 10))
println("длины: ", length(v_1), ", ", length(v_2), ", ", length(v_3))
println("сумма = ", v_sum)
```
