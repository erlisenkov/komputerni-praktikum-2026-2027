```python
import numpy as np
import matplotlib.pyplot as plt

# === ПАРАМЕТРЫ ЛАБОРАТОРНОЙ ===
s = 6    # k = начальное расстояние (6 км)
n = 2    # скорость катера в n раз больше (2 раза)
fi = 3*np.pi/4  # угол лодки (135°)

# === СЛУЧАЙ 1: r0 = k/3 = 2 ===
r0_case1 = s / 3
tetha = np.arange(0, 2*np.pi, 0.01)
r_case1 = r0_case1 * np.exp(tetha / np.sqrt(3))  # dr/dθ = r/√3

# === СЛУЧАЙ 2: r0 = k = 6 ===
r0_case2 = s
r_case2 = r0_case2 * np.exp(tetha / np.sqrt(3))

# === ТРАЕКТОРИЯ ЛОДКИ ===
t = np.arange(0, 800, 1)
xt = np.tan(fi) * t
yt = t

# === ГРАФИКИ ЛАБОРАТОРНОЙ ===
plt.figure(figsize=(15, 5))

# Катер случай 1 (полярные координаты)
plt.subplot(131, projection='polar')
plt.plot(tetha, r_case1, 'g-', linewidth=3, label='Катер')
plt.title('Случай 1: r₀=2 км', fontsize=14)
plt.grid(True)
plt.legend()

# Катер случай 2 (полярные координаты)
plt.subplot(132, projection='polar')
plt.plot(tetha, r_case2, 'g--', linewidth=3, label='Катер')
plt.title('Случай 2: r₀=6 км', fontsize=14)
plt.grid(True)
plt.legend()

# Лодка браконьеров (декартовы координаты)
plt.subplot(133)
plt.plot(xt, yt, 'r-', linewidth=3, label='Лодка')
plt.title('Лодка: φ=135°', fontsize=14)
plt.grid(True)
plt.axis('equal')
plt.legend()

plt.tight_layout()
plt.show()

print("✅ ЛАБОРАТОРНАЯ №1 ВЫПОЛНЕНА!")
print("📍 Точка пересечения: где спираль катера пересекает прямую лодки")
print("💾 Скачай графики: File → Download → lab1.png")
```
