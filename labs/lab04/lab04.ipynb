```julia
A = rand(3, 3)
x = fill(1.0, 3)
b = A*x
println("Решение СЛАУ A\\b: ", A\b)

Alu = lu(A)
println("LU-факторизация, матрица L:\n", Alu.L)

Aqr = qr(A)
println("QR-факторизация, матрица Q:\n", Aqr.Q)

Asym = A + A'
AsymEig = eigen(Asym)
println("Собственные значения симметричной матрицы: ", AsymEig.values)
```

```julia
# 1. Диагонализация
A1 = [1 -2; -2 1]
E1 = eigen(A1)
println("Диагональный вид A1: ", Diagonal(E1.values))

# 2. Степени и корни
println("A1^10:\n", A1^10)

A2 = [5 -2; -2 5]
println("Квадратный корень из A2:\n", sqrt(A2))

A3 = [1 -2; -2 1]
E3 = eigen(A3)
# ИСПРАВЛЕНО: cbrt работает с отрицательными числами (вещественный кубический корень)
cbrt_A3 = E3.vectors * Diagonal(cbrt.(E3.values)) / E3.vectors
println("Кубический корень из A3:\n", cbrt_A3)
println("Проверка (cbrt_A3)^3:\n", cbrt_A3^3)

A4 = [1 2; 2 3]
E4 = eigen(A4)
# ИСПРАВЛЕНО: у A4 есть отрицательное собственное значение,
# поэтому вещественного корня нет — считаем комплексный через спектральное разложение
sqrt_A4 = E4.vectors * Diagonal(sqrt.(Complex.(E4.values))) / E4.vectors
println("Квадратный корень из A4 (комплексный):\n", sqrt_A4)
println("Проверка sqrt_A4^2:\n", sqrt_A4^2)

# 3. Собственные значения матрицы 5x5
A5 = [140 97 74 168 131;
      97 106 89 131 36;
      74 89 152 144 71;
      168 131 144 54 142;
      131 36 71 142 36]

E5 = eigen(A5)
println("Собственные значения A5: ", E5.values)
D5 = Diagonal(E5.values)
L5 = tril(A5)
println("Нижнетреугольная матрица из A5:\n", L5)
```

```julia
function check_productive(A)
    I_mat = Matrix{Float64}(I, size(A))
    
    # Критерий 1: Все элементы (I - A)^-1 >= 0
    inv_IA = inv(I_mat - A)
    crit1 = all(inv_IA .>= -1e-10) # с учетом погрешности вычислений
    
    # Спектральный критерий: Все собственные значения по модулю < 1
    eig_vals = eigvals(A)
    crit2 = all(abs.(eig_vals) .< 1.0)
    
    return crit1, crit2
end

# Примеры матриц (замени на свои из варианта, если они отличаются)
A_test1 = [0.1 0.2; 0.3 0.4]
A_test2 = [0.5 0.2; 0.1 0.3]

c1, c2 = check_productive(A_test1)
println("Матрица 1 продуктивна? Критерий (I-A)^-1: ", c1, " | Спектральный: ", c2)

c1, c2 = check_productive(A_test2)
println("Матрица 2 продуктивна? Критерий (I-A)^-1: ", c1, " | Спектральный: ", c2)
```
