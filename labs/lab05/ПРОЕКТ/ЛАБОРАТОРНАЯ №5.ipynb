```julia
pyplot()
sin_theor(x) = sin(x)
sin_taylor(x) = [(-1)^i * x^(2*i+1) / factorial(2*i+1) for i in 0:4] |> sum

plot(sin_theor, -5, 5, title="График sin(x)", xlabel="x", ylabel="y", label="sin(x)")
plot(sin_taylor, -5, 5, title="Ряд Тейлора", xlabel="x", ylabel="y", label="Тейлор")

plot(sin_theor, -5, 5, label="sin(x)")
plot!(sin_taylor, -5, 5, label="Тейлор")

plot(sin_taylor, -5, 5,
     label="sin(x), разложение в ряд Тейлора",
     line=(:blue, 0.3, 6, :solid),
     size=(800, 500),
     xticks=(-5:0.5:5), yticks=(-1:0.1:1),
     xtickfont=font(12, "Times New Roman"),
     ytickfont=font(12, "Times New Roman"),
     ylabel="y", xlabel="x",
     title="Разложение в ряд Тейлора",
     xrotation=rad2deg(pi/4),
     fillrange=0, fillalpha=0.5, fillcolor=:lightgoldenrod,
     background_color=:ivory)
plot!(sin_theor, -5, 5, label="sin(x), теоретическое значение",
      line=(:black, 1.0, 2, :dash))

savefig("taylor.pdf")
savefig("taylor.png")
```

```julia
r_h = 1; k_h = 3; n_h = 100
θh = collect(0:2π/100:2π + 2π/100)
Xh = r_h * k_h * cos.(θh)
Yh = r_h * k_h * sin.(θh)

anim_h = @animate for i in 1:n_h
    plt = plot(5, xlim=(-4, 4), ylim=(-4, 4), c=:red, aspect_ratio=1,
               legend=false, framestyle=:origin)
    plot!(plt, Xh, Yh, c=:blue, legend=false)
    t = θh[1:i]
    xh = r_h*(k_h-1)*cos.(t) + r_h*cos.((k_h-1)*t)
    yh = r_h*(k_h-1)*sin.(t) - r_h*sin.((k_h-1)*t)
    plot!(xh, yh, c=:red)
    xc = r_h*(k_h-1)*cos(t[end]) .+ r_h*cos.(θh)
    yc = r_h*(k_h-1)*sin(t[end]) .+ r_h*sin.(θh)
    plot!(xc, yc, c=:black)
    xl = transpose([r_h*(k_h-1)*cos(t[end]) xh[end]])
    yl = transpose([r_h*(k_h-1)*sin(t[end]) yh[end]])
    plot!(xl, yl, markershape=:circle, markersize=4, c=:black)
    scatter!([xh[end]], [yh[end]], c=:red, markerstrokecolor=:red)
end
gif(anim_h, "hypocycloid.gif", fps=15)
```

```julia
using Statistics
sds = [1, 1/2, 1/4, 1/8, 1/16, 1/32]
n_e = 10
y_e = [mean(sd * randn(n_e)) for sd in sds]
errs = 1.96 .* sds / sqrt(n_e)

plot(y_e, ylims=(-1, 1), title="Исходные значения", label="y", marker=:circle)
plot(y_e, ylims=(-1, 1), err=errs, title="Значения с ошибками", label="y ± err", marker=:circle)
plot(y_e, 1:length(y_e), xerr=errs, marker=stroke(3, :orange),
     title="Повёрнутый график с ошибками", label="y")
plot(y_e, ribbon=errs, fill=:cyan, title="Заливка области ошибок", label="y")
```

```julia
function cycloid_anim(r, k, fname, is_epi::Bool)
    q = denominator(k)
    T = 2π * q
    n = 200
    ts = range(0, stop=T, length=n)
    R = r * k
    anim = @animate for i in 1:n
        plt = plot(xlim=(-2.2R, 2.2R), ylim=(-2.2R, 2.2R), aspect_ratio=1,
                   legend=false, framestyle=:origin, title="k = $(k)")
        plot!(plt, R*cos.(ts), R*sin.(ts), c=:blue)
        t = ts[1:i]
        if is_epi
            xx = r*(k+1)*cos.(t) - r*cos.((k+1)*t)
            yy = r*(k+1)*sin.(t) - r*sin.((k+1)*t)
        else
            xx = r*(k-1)*cos.(t) + r*cos.((k-1)*t)
            yy = r*(k-1)*sin.(t) - r*sin.((k-1)*t)
        end
        plot!(xx, yy, c=:red, lw=2)
    end
    gif(anim, fname, fps=30)
end

cycloid_anim(1.0, 3//1, "hypo_k3.gif", false)
cycloid_anim(1.0, 4//1, "hypo_k4.gif", false)
cycloid_anim(1.0, 5//2, "hypo_k5_2.gif", false)
cycloid_anim(1.0, 7//3, "hypo_k7_3.gif", false)
cycloid_anim(1.0, 3//1, "epi_k3.gif", true)
cycloid_anim(1.0, 4//1, "epi_k4.gif", true)
cycloid_anim(1.0, 5//2, "epi_k5_2.gif", true)
cycloid_anim(1.0, 7//3, "epi_k7_3.gif", true)
```
